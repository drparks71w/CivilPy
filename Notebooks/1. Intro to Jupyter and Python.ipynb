### STRINGS

#### What is a String?
A string is a data type that is used to represent text rather than numbers and is composed of a set of characters and can also contain spaces and numbers.

#### Why do we use Strings?
Strings are useful when storing small or large bodies from single words to full sentences. 

#### How do we use Strings?
Strings are assigned to variables using a single equals and are surrounded by triple quotes `""" """`, double quotes `" "` or single quotes `' '`, although triple quotes are generally used for function doc strings so please use the single or double quotes.

#### Creating a String

````python
string_1 = """This is a string"""  # triple quote
string_2 = "this is another string"  # double quote
string_3 = "this is also a string"  # single quote

print(string_3)
type(string_1)
````

#### Indexing a String

It is possible to index a string by referencing the position (or index) of a single character, or by referencing a range with a start and end position. Note: indexes start at where the first element is 0.

````python
print(string_1)
print(string_1[0])
print(string_1[5:-3])
````

#### Formatting Strings (f-strings)

Prefixing a string with the letter `f` lets you drop variables - or any python expression - directly into the text by wrapping them in curly braces. This is the standard way to produce readable output, and you'll use it constantly when printing calculation results. A format specifier after a colon controls how numbers display: `:.1f` rounds to 1 decimal place, `:,` adds thousands separators.

````python
girder = "W36x150"
span = 92.5

print(f"Girder {girder} spans {span} ft")
print(f"That's {span * 12} inches")
print(f"Max moment: {1250.4567:.1f} kip-ft")
print(f"Total load: {1_234_567:,} lbs")
````

### INTEGERS or INTs

#### What is an Integer?

An integer is a whole number (not a fraction) that can be positive, negative or zero.

#### Why do we use Integers?

Integers are a commonly used data type in programming as they are efficient to store and are generally used in loops and indexing strings or lists. 

#### How do we use Integers?

Integers are assigned to variables using a single equals and are standalone numbers written without a decimal point. Be aware that the division operator `/` **always** returns a float in python, even when the result is a whole number - use `//` (floor division) when you want an integer back.

````python
int_1 = 1
int_2 = -2
int_3 = 10 / 2  # careful - the / operator always returns a float
int_4 = 10 // 2  # floor division returns an int

print(int_3, type(int_3))
print(int_4, type(int_4))
````

### FLOATING-POINT NUMBERS or FLOATS

#### What is a Float?

A float is a floating-point number that contains a decimal point. Compared to Integers, they require more space in memory as you will have two integers separated by a decimal point.

#### Why do we use Floats?

Computers recognize real numbers that contain fractions as floating point numbers. Floats are useful for numbers that require a great deal of precision, such as calculations in engineering or microchip manufacturing.

#### How do we use Floats?

Floats are assigned to variables using a single equals and are standalone numbers that contain a decimal point or the result of an operation that returns a decimal number.

````python
float_1 = 1.0
float_2 = 3 / 1
print(float_1, float_2)
type(float_1)
````

### BOOLEANS

#### What is a Boolean?

A boolean is a data type with only two possible values: `True` or `False` (capitalization matters).

#### Why do we use Booleans?

Booleans are the yes/no answers of programming - they're what makes decision making possible. Comparison operators (`==` equal, `!=` not equal, `<`, `>`, `<=`, `>=`) all return booleans, and the `if` statements covered in the Control Flow section below run code based on them.

#### How do we use Booleans?

Booleans are either assigned directly or, far more commonly, produced by a comparison. They can be combined with `and`, `or` and `not`.

````python
is_composite = True

girder_depth = 36
print(girder_depth > 30)
print(girder_depth == 36, girder_depth != 36)
print(girder_depth > 30 and not is_composite)
type(is_composite)
````

### LISTS

#### What is a List?

Lists are data structures that contain an ordered collection of elements. These elements are generally of the same type and can contain elements such as single data types through to more complex data structures such as other lists or dictionaries.

#### Why do we use Lists?
Lists are commonly used in programming to store and organize data so it can be easily sorted, indexed or searched.

#### How do we use Lists?

Lists are assigned to variables using a single equals, are denoted by square brackets, and use a comma to separate elements `[a,b]`.

#### Creating a List

There are two main ways of creating a list. First is to use square brackets,`a=[]` and the second is to use the list type cast `b=list()`.

````python
list_1 = ["string item", float_1, 100]
list_a = list()
print(list_1)
type(list_1)
````

#### Adding an item to a List

Now that we have our list, you can add another element using the python's inbuilt `.append()` method.

````python
list_1.append("new element")
print(list_1)
type(list_1)
````

#### Changing an item in a List

When changing an item in a list, you need to know the position of the item denoted by the index.

````python
list_1[1] = "Changed Item"

print(list_1)
````

#### Removing an item from a List

To remove an item from a list, use the `.pop()` method.

````python
list_1.pop(0)
print(list_1)
````

#### Merging two Lists

There are two ways you can merge two lists. First is to append using `.append(element)` which will add the new elements to the end of the old list.

````python
list_2 = ["this", "is", "a", "list"]
list_1.append(list_2)
print(list_1)
````

The other way is to insert a list as an element within the list (note: not append but inject) using the `.insert(index, element)` method.

````python
list_3 = [1, 2, 3, 4]
list_4 = [5, 6, 7, 8]
list_3.insert(2, list_4)
print(list_3)
````

#### Accessing items in a List ###
You can access list items by referring to the index number of the item. Note: Lists start with an index of 0. 

You can also use negative indexing to access items from back to front. E.g. -1 is the last item in the list and -2 is second last.

````python
print(list_1[0])
print(list_1[-1])
````

You can also reference index ranges by using the colon between indexes which refer to the start and end of the range e.g. `[start:end]` or if you omit a number on either start (or end), it will include everything from the start (or end) up until the end (or start) e.g. `[:end]` will include the first item until the end index item.

````python
print(list_1[0:3])
print(list_1[2:])
````

### Dictionaries

#### What is a Dictionary?

Dictionaries are data structures that contain a collection of "key" and "value" pairs. Each key must be unique - keys are usually strings, though any immutable type works - while the value can be anything from a single variable to another data structure such as a list or a second dictionary. Note that assigning to a key that already exists silently **overwrites** the old value; python won't raise an error or warn you, so be deliberate when reusing keys.

#### Why do we use a Dictionary?

Dictionaries are commonly used in programming to store and organize data so it can be easily sorted, indexed or searched. Since dictionaries are mutable, we can add, change, or remove elements.

#### How do we use a Dictionary?

## CONTROL FLOW

So far we've only *stored* data. Control flow is where python starts doing work for you: `if` statements make decisions, and loops repeat an operation over every item in a collection. If you only take two things away from this notebook, make it these two - nearly every "automate the boring stuff" task is a loop wrapped around a few if statements.

### Conditionals: if / elif / else

An `if` statement runs its indented block only when its condition is `True`. `elif` ("else if") chains additional checks, and `else` catches everything that fell through. Indentation is not optional in python - the indented lines are what define which code belongs to the statement.

````python
span_ft = 120

if span_ft <= 40:
    bridge_type = "concrete slab"
elif span_ft <= 100:
    bridge_type = "prestressed box beam"
else:
    bridge_type = "steel plate girder"

print(f"A {span_ft} ft span suggests a {bridge_type} bridge")
````

### For Loops

A `for` loop repeats a block of code once for every item in a list (or any other *iterable* - strings, dictionaries and ranges all work). This is the single most useful concept in this notebook: any time you catch yourself doing the same operation on 40 girders, 200 borings, or 1,000 site photos, a for loop should be doing it instead.

````python
span_lengths = [85.5, 120.0, 96.0, 110.5]  # ft

for span in span_lengths:
    print(f"{span} ft = {span * 12} in")
````

A common pattern is accumulating a result as you loop. Also useful: `range(n)` generates the numbers `0` through `n-1` when you just need a counter;

````python
total_length = 0
for span in span_lengths:
    total_length = total_length + span

print(f"Total bridge length: {total_length} ft")

for i in range(3):
    print(i)
````

### While Loops

A `while` loop repeats *as long as* its condition stays `True` - useful when you don't know the number of iterations in advance. Make sure something inside the loop eventually makes the condition `False`, or it will run forever (if that happens, the stop &#9632; button in the toolbar interrupts the kernel).

````python
deck_thickness = 8.0  # inches

# thicken the deck in half inch steps until a 1 in strip provides I >= 100 in^4
while deck_thickness**3 / 12 < 100:
    deck_thickness = deck_thickness + 0.5

print(f"Required thickness: {deck_thickness} in")
````

### Comprehensions

Now that you've seen for loops, *comprehensions* will make sense: they are a one-line shorthand for building a list (or dictionary) from a loop.

#### List Comprehension

List comprehensions provide a concise way to create lists where each new element is the result of some operation applied to each member of another sequence or iterable. 

````python
squares = [x**2 for x in range(10)]
print(squares)
type(squares)
````

#### Dictionary Comprehension

## Searching the Full AISC Shapes Table

Creating sections one label at a time is fine when you already know what you want, but civilpy also exposes the entire AISC shapes database as a pandas `DataFrame` - a spreadsheet-like table you can filter, sort and search. This is how you answer questions like *"what's the lightest W shape with a section modulus of at least 500 in³?"* without flipping through the steel manual.

````python
from civilpy.structural.steel import steel_tables

print(steel_tables.shape)  # (rows, columns)
steel_tables.head(3)
````

Each row is a shape and each column is a property straight out of the AISC database (`W` = weight in plf, `A` = area, `Ix`/`Iy` = moments of inertia, `Sx`/`Zx` = elastic/plastic section moduli, and so on). Filtering uses the pattern `table[condition]`. Below, every W shape with $S_x \geq 500\ \text{in}^3$, sorted so the lightest option is on top;

````python
w_shapes = steel_tables[steel_tables["Type"] == "W"]

candidates = w_shapes[w_shapes["Sx"] >= 500]
candidates.sort_values("W")[["EDI_Std_Nomenclature", "W", "A", "d", "Ix", "Sx", "Zx"]].head()
````

The lightest section that works is a W40X149 - assuming depth isn't limited. Multiple conditions combine with `&` (and) or `|` (or), each wrapped in parentheses. The same idea searching on moment of inertia, with a weight cap;

````python
candidates = w_shapes[(w_shapes["Ix"] >= 9000) & (w_shapes["W"] <= 160)]
candidates.sort_values("W")[["EDI_Std_Nomenclature", "W", "A", "d", "Ix", "Sx", "Zx"]]
````
